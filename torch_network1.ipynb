### Output Layer Delta

**Formula:**

$$
\delta^L = \nabla_a C \odot \sigma'(z^L)
$$

**Step 1: Loss gradient**

$$
\nabla_a C = \frac{\partial C}{\partial a^L}
$$

Measures how the loss changes with the output activation.

**Step 2: Activation derivative**

$$
\sigma'(z^L) = \frac{\partial a^L}{\partial z^L}
$$

Measures activation sensitivity.

**Step 3: Calculate delta**

$$
\delta^L = \frac{\partial C}{\partial a^L}
\odot
\frac{\partial a^L}{\partial z^L}
$$

Combines both to calculate the output layer's error signal.

**Purpose:** Measures how changes in the output neuron's weighted input affect the loss.
