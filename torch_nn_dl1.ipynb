```python
class LinearModel(nn.Module):#Class inheritance from module not Linear Model,Meta class object

  #in features,out features =>input and output neurons
  def __init__(self
               ,in_features,
               out_features):
    #!calling parent/module class object
    super().__init__()
    #!defining linear attribute as object of a linear layer object
    self.linear=nn.Linear(in_features=in_features,
                          out_features=out_features)

  #forward object function
  #linear has already passed in and out,passing x now
  #!linear had these parameters already in,out and x
  def forward(self,x):
    y_pred=self.linear(x)
    return y_pred
  #requires_grad=True for the object/tensor

```

```python
#inherit from module(layer holder)
#!define init as input and output layer
#intitated module
#!define linear attribute with linear dense layer(feed forward)
#!forward function=>input output already defined size wise pass x
```

```python
criterion=nn.MSELoss()
optmizer=torch.optim.SGD(linear_model_L2.parameters(),lr=0.001)
```

```python
#epoch is a single pass through all the layers for the entire dataset/input
```

```python
X
```

```python
epochs=50
loss_history=[]
for i in range(50):
  i+=1
  y_pred=linear_model_L2.forward(X)
  loss=criterion(y_pred,Y)
  loss_history.append(loss.item())
  print(f"Epoch:{i}: {loss.item()}  weight {linear_model_L2.linear.weight.item()} {linear_model_L2.linear.bias.item()} ")
  optmizer.zero_grad()#fresh data on each epoch and no accumulation of gradent,weights keep getting better often
  loss.backward()#backprop
  optmizer.step()#weigh bias adjust

#loo
#     loss compute
#!    optmizer reset
#!    loss.backprop
#!    optmizer.step()
```
